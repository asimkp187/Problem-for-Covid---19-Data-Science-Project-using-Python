# 6. Date time format :

# 7. Data Aggregation:
#### a. Find max value in all columns using groupby function on 'continent' column

```python
x.groupby('continent').max().reset_index()
```

#### b. Store the result in a new dataframe named 'df_groupby'.

```python
x_groupby=x.groupby('continent').max().reset_index()
x_groupby
```

# 8. Feature Engineering :
#### Create a new feature 'total_deaths_to_total_cases' by ratio of'total_deaths' column to 'total_cases'

```python
x_groupby["total_deaths_to_total_cases"]=x_groupby["total_deaths"]/x_groupby["total_cases"]
```

```python
x_groupby["total_deaths_to_total_cases"]*100
```

```python
x_groupby
```
